```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt




url = 'https://raw.githubusercontent.com/cpethe/TED_Talks/master/ted_main.csv'
ted_data = pd.read_csv(url)

filters= [100, 500, 1000, None]
filter_names= ['at least 100 comments', 'at least 500 comments', 'at least 1000 comments', 'all comments']
plt.figure(figsize=(15, 10))
for i , min_comments in enumerate (filters):
  if (min_comments):
    # this if condition is reached for all filters except the 'none'
    subset=ted_data[ted_data['comments'] >=min_comments]
  else:
    subset= ted_data

#log of 0 is undefined, so we add one when calculating
  subset['log_comments']= np.log(subset['comments']+1)
  subset['log_views']= np.log (subset['views']+1);

  plt.subplot(2,2,i+1)
  plt.scatter(subset['log_comments'], subset['log_views'], alpha=0.4, s=10)
  plt.title(filter_names[i])
  plt.xlabel('Log(Comments)')
  plt.ylabel('Log(Views)')
  plt.grid(True)

  plt.tight_layout()
  plt.show()










```

I found that the most revealing point size was 10, with a transparency of 0.4.


```python

import matplotlib.pyplot as plt
import numpy as np

# Filter data with at least 750 comments
filtered_data = ted_data[ted_data['comments'] >= 750]

# Log transformation
filtered_data['log_comments'] = np.log(filtered_data['comments'] + 1)
filtered_data['log_views'] = np.log(filtered_data['views'] + 1)
filtered_data['log_duration'] = np.log(filtered_data['duration'] + 1)

# Initialize the 3D plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')


color_map = 'plasma'
sc = ax.scatter(filtered_data['log_comments'], filtered_data['log_views'], filtered_data['log_duration'],
                c=filtered_data['log_duration'], cmap=color_map)

ax.set_xlabel('Log(Comments)')
ax.set_ylabel('Log(Views)')
ax.set_zlabel('Log(Duration)')
plt.colorbar(sc, label='Log(Duration)')
plt.title(f"3D Scatter Plot with {color_map} Color Map")
plt.show()
```

```python
import seaborn as sns

# Plot the violin plot
plt.figure(figsize=(10, 8))
sns.violinplot(x=ted_data['duration'])
plt.xlabel('Duration')
plt.title('Violin Plot of Duration')
plt.grid(True)
plt.show()
```

```python


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


olympics_data = pd.read_csv('https://www.data-manual.com/datafiles/olympics.csv')
olympics_data.head()

# Select only numeric columns
numeric_columns = olympics_data.select_dtypes(include=['number']).columns

# (z-score normalization)
normalized_data = (olympics_data[numeric_columns] - olympics_data[numeric_columns].mean()) / olympics_data[numeric_columns].std()


normalized_data = normalized_data.set_index(olympics_data['Country name'])



plt.figure(figsize=(12, 8))
sns.heatmap(normalized_data, cmap='coolwarm', annot=True)
plt.title('2012 Olympics Data Heatmap')
plt.xlabel('Features')
plt.ylabel('Country')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


olympics_data = pd.read_csv('https://www.data-manual.com/datafiles/olympics.csv')
olympics_data.head()

medals_data = olympics_data[['Country name', 'Gold medals', 'Silver medals', 'Bronze medals']]
medals_data = medals_data.set_index('Country name')

# Plot grouped bar chart
medals_data.plot(kind='bar', stacked=False, figsize=(12, 8), color=['gold', 'silver', 'brown'])
plt.title('Olympic Medals by Country')
plt.xlabel('Country')
plt.ylabel('Number of Medals')
plt.legend(title='Medal Type')
plt.show()

```
