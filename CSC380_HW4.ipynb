```python
# code for Q5 calculations
#used SI Session 7 code as reference


import numpy as np
import scipy.stats as stats
# Step 1: Observed frequencies
observed = np.array([[18,26],[8,48]])
# Step 2: Perform chi-square test
chi2_stat, p_val, dof, expected = stats.chi2_contingency(observed)
# Step 3: Print the results
print("Observed Frequencies:")
print(observed)
print("\nExpected Frequencies:")
print(np.round(expected, 4))
print("\nChi-square Statistic:", round(chi2_stat, 4))
print("Degrees of Freedom:", dof)
print("P-value:", round(p_val, 4))
# Step 4: Check the critical value at 0.05 significance level
alpha = 0.05
critical_value = stats.chi2.ppf(1 - alpha, dof)
print("\nCritical Value at 0.05 significance level:", round(critical_value, 4))
# Step 5: Conclusion
if chi2_stat > critical_value:
  print("Reject the null hypothesis. There is a significant association between a person’s age and whether they actively use social media.")

else:
  print("Fail to reject the null hypothesis. There is no significant association between a person’s age and whether they actively use social media.")














```

```python
import numpy as np
import pandas as pd



url = 'https://raw.githubusercontent.com/cpethe/TED_Talks/master/ted_main.csv'
ted_data = pd.read_csv(url)

############################

occupation_count=ted_data['speaker_occupation'].value_counts()
common_occupation=occupation_count.idxmax()
common_count=occupation_count.max()


print(f"The most common speaker occupation is '{common_occupation}' with {common_count} occurrences.")

##################################

# Drop the 'title' column
ted_data = ted_data.drop(columns=['name'])

ted_data.head()
###################################
```

```python
cc_in_descr = ted_data["description"].apply(lambda descr_str: "climate change" in descr_str)
ted_data[cc_in_descr]
```

```python

filtered_talks = ted_data[(ted_data['views'] >= 10_000_000) | (ted_data['comments'] >= 3000)]
sorted_talks = filtered_talks.sort_values(by='duration', ascending=True)

top_10_longest_talks = sorted_talks.head(10)

print(top_10_longest_talks)
```

```python
from ast import literal_eval


# Initialize variables to track the talk with the max difference in views
max_diff_talk = None
max_diff_value = 0

# Iterate through each row to calculate view differences
for index, row in ted_data.iterrows():

    if pd.notna(row['related_talks']):
        related_talks = literal_eval(row['related_talks'])


        # Calculate the view difference for each related talk
        for related in related_talks:

            view_difference = abs(row['views'] - related['viewed_count'])

            # Check if this is the maximum difference encountered
            # if not, we switch it to curr
            if view_difference > max_diff_value:
                max_diff_value = view_difference
                max_diff_talk = row

# Display the talk with the largest view difference
print("Talk with the maximum view difference compared to related talks:")
print(max_diff_talk[['title']])
print(f"Maximum difference in views: {max_diff_value}")
```

```python
# Select only numeric columns for correlation
numeric_columns = ted_data.select_dtypes(include=['float64', 'int64'])
correlation_matrix = numeric_columns.corr()
#now, we convert the 2-d into a 1-d version to make it easier to compute further
correlation_pairs = correlation_matrix.unstack()
sorted_pairs = correlation_pairs.sort_values(ascending=False) # sorting

#we have self-pairs with a correlation of 1 , so we get rid of those
sorted_pairs = sorted_pairs[sorted_pairs < 1]

#getting max
# Get the most correlated pair
most_correlated = sorted_pairs.idxmax()
most_correlated_value = sorted_pairs.max()
print(f"Most correlated Features: {most_correlated}, Correlation: {most_correlated_value}")

```

```python
#The p-value indicates the probability that the observed correlation is due to chance.
#the Pearson correlation test, which provides both the correlation coefficient and a p-value, is a good one because
# The p-value indicates whether the observed correlation is statistically significant, depending on the
# t statistic used
# this can help us understand the correlation between the elements of the csv being assessed
from scipy.stats import pearsonr

# Drop rows with missing values in 'duration' or 'comments'
ted_data = ted_data.dropna(subset=['duration', 'comments'])
# Calculate the Pearson correlation coefficient and p-value
correlation_coefficient, p_value = pearsonr(ted_data['duration'], ted_data['comments'])

#f strings to display our found correlation
print(f"Correlation Coefficient: {correlation_coefficient}")
print(f"P-Value: {(p_value)}")

```
